**50. Find the total revenue using np.dot on the hours array and the rate array.**

```python
valid_entries = ~np.isnan(hours) & ~np.isnan(billing_rate)

hours_valid = hours[valid_entries]
rate_valid = billing_rate[valid_entries]

#total_revenue = np.dot(hours, billing_rate)
total_revenue = np.dot(hours_valid, rate_valid)

print(total_revenue)
```

Calculated the total revenue as product of billing rate and number of hours logged for billing.
