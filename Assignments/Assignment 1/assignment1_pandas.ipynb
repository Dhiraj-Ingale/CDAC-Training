**25. Convert the hours logged column to a number.**

```python
df['hours_logged'] = pd.to_numeric(df['hours_logged'].str.replace('h', ''), errors='coerce')

df
```

.to_numeric() method is used to convert to number data type.

**26. Convert the billing rate column to a number. Remove the $ sign first.**

```python
df['billing_rate_(usd)'] = df['billing_rate_(usd)'].astype(str).str.replace('$', '', regex=False).str.strip()

df['billing_rate_(usd)'] = pd.to_numeric(df['billing_rate_(usd)'], errors='coerce')

df
```

To convert billing rate to number first need to remove $ symbol from entries.

**27. Convert the CSAT score column to a number.**

**34. Add a new column that holds the year of the ticket.**

**35. Add a new column that holds the month of the ticket.**

```python
df['created_month'] = df['ticket_created_at'].dt.strftime("%B")

df
```
